# 4. Gaussian Process Regression

## Iterative GPR

Designing a MEMS backplate requires tuning its fundamental resonance frequency $f_0$ while balancing acoustic performance, mechanics, and fabrication limits. Because FEM simulations are computationally expensive, we use Gaussian Process Regression (GPR) as a surrogate model to directly learn the forward mapping from geometry to frequency:

$$(a, h, p, d) \;\xrightarrow{\quad\mathcal{GP}\quad}\; f_0$$

**Design Parameters & Fabrication Constraints**

The plate geometry is governed by four continuous parameters:
* $a$: Plate half-side length
* $h$: Plate thickness
* $p$: Perforation pitch (hole-to-hole center distance)
* $d$: Perforation hole diameter

A viable microphone plate must satisfy strict physical validity and fabrication rules:

* Thin-Plate Assumption ($h/p < 2$)
* Perforation Ratio ($R = (d/p)^2 < 0.5$)
* Lithographic Clearance ($p \ge d + 5\text{ um}$): Preserves a minimum sidewall ligament of $5\text{ um}$ between adjacent etch holes to prevent sidewall collapse during release etching.

![Geometry](geom.png)


You can play around with setting of variables `N_TRAIN` and `N_CANDIDATES`:

1. `N_TRAIN` (High-Fidelity Physics Evaluations)
* Meaning: The number of calls made to the ground-truth physical solver.

* Computational Cost: EXTREMELY EXPENSIVE. In real engineering workflows, each run  represents a 3D finite-element solve taking minutes to hours.

* Engineering Goal: Keep N_TRAIN as LOW as possible to minimize total solve time.

2. `N_CANDIDATES` (Surrogate Query Pool)
* Meaning: The number of virtual geometry designs evaluated ONLY by the trained Gaussian Process surrogate model (gp.predict).

* Computational Cost: VIRTUALLY FREE. Running vectorized matrix math over 50,000 points takes a fraction of a second in NumPy/scikit-learn. No physics solver is executed.

* Engineering Goal: Keep N_CANDIDATES HIGH (50,000 to 100,000). A dense candidate cloud provides fine geometric resolution across the 4D parameter space (a, h, p, d), allowing instant inversion and target lookup via np.argmin without running numerical optimizers.

**Goal:** Your task is to answer *How Low Can We Go?* and minimize the number of calls of computationally expensive simulation calls `get_freq()`.




```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, ConstantKernel as C

from gp_utils import get_freq
```

### Configuration and Parameter Bounds

```python
TARGET_FREQ = 45000.0  # target fundamental frequency in Hz
target_khz = TARGET_FREQ / 1e3

BOUNDS = np.array([
    [0.5e-3, 2.0e-3],   # a: plate half-side (m)
    [5.0e-6, 20.0e-6],  # h: thickness (m)
    [20.0e-6, 50.0e-6], # p: pitch (m)
    [2.0e-6, 10.0e-6]   # d: diameter (m)
])

lower_b = BOUNDS[:, 0]
upper_b = BOUNDS[:, 1]

def to_unit(x):
    """Normalize physical parameters into [0, 1]."""
    return (x - lower_b) / (upper_b - lower_b)

def from_unit(u):
    """Unscale unit coordinates back to physical units."""
    return lower_b + u * (upper_b - lower_b)

def sample_valid_designs(n_samples, rng):
    # Samples n_init points uniformly across the parameter box [a, h, p, d] 
    # using random scaling between the lower and upper bounds (rng)

    u = rng.rand(n_samples, 4)
    x = from_unit(u)
    # Ensure R = (d/p)^2 < 0.5
    x[:, 3] = np.minimum(x[:, 3], x[:, 2] * np.sqrt(0.49))
    return x
```

### Initial Dataset Generation and GPR Setup


```python
# ==============================================================================
# Hyperparameters & Convergence Criteria
# ==============================================================================
# TODO play around with these numbers:
N_INIT = 10              # Initial batch of expensive physics calls
MAX_ITER = 25            # Maximum active learning query loops
N_CANDIDATES = 50000     # Virtual candidate pool evaluated via surrogate
STOPPING_TOL_PCT = 1     # Stop if relative error <= 1%

# Fix random seed for deterministic sampling and kernel optimization
rng = np.random.RandomState(42)

# ==============================================================================
# Initial Design Generation & Physics Evaluation
# ==============================================================================
# Generate initial N_INIT points within bounds
X_train_phys = sample_valid_designs(N_INIT, rng)

# Evaluate exact fundamental natural frequency f_0 (scaled to kHz for numerical stability)
y_train_khz = np.array([
    get_freq(row[0], row[1], row[2], row[3], m=1, n=1) / 1e3
    for row in X_train_phys
])

# ==============================================================================
# Gaussian Process Fit (Normalized Design Space [0, 1]^4)
# ==============================================================================
# Initialize and fit standard GP on the first 10 points
kernel = C(1.0, (1e-2, 1e4)) * Matern(length_scale=[0.5, 0.5, 0.5, 0.5], nu=2.5)
gp = GaussianProcessRegressor(
    kernel=kernel,
    alpha=1e-4,                 # Tikhonov regularization (nugget) for numerical stability
    n_restarts_optimizer=5,     # Multiple restarts to avoid local L-BFGS-B minima in MLE
    random_state=42,
    normalize_y=True, # the GP's prior is now set to sample mean of the training data, which can help with convergence and stability - try to turn it off and see what happens
)
gp.fit(to_unit(X_train_phys), y_train_khz)


# ==============================================================================
# Candidate Space & Surrogate Inversion
# ==============================================================================
# Generate a dense candidate pool and project into normalized coordinates
candidates_phys = sample_valid_designs(N_CANDIDATES, rng)
candidates_u = to_unit(candidates_phys)

# Compute posterior mean (mu) and standard deviation (sigma) across all candidates
init_pred_f, init_pred_std = gp.predict(candidates_u, return_std=True)

# Select the candidate whose posterior mean is closest to the target frequency
init_best_idx = np.argmin(np.abs(init_pred_f - target_khz))
init_pred_target_khz = init_pred_f[init_best_idx]
init_pred_target_std_khz = init_pred_std[init_best_idx]

# Ground-Truth Validation of the Initial Surrogate Recommendation
init_best_params = candidates_phys[init_best_idx]
init_true_freq_khz = get_freq(*init_best_params, m=1, n=1) / 1e3
init_true_rel_error = abs(init_true_freq_khz - target_khz) / target_khz * 100.0

print("Based on the initial surrogate model, the best candidate is:")
print(f"  Plate Half-side a:    {init_best_params[0]*1e6:.3f} um")
print(f"  Thickness h:          {init_best_params[1]*1e6:.2f} um")
print(f"  Pitch p:              {init_best_params[2]*1e6:.2f} um")
print(f"  Diameter d:           {init_best_params[3]*1e6:.2f} um")
print(f"  Surrogate's guess:       {init_pred_target_khz:.3f} +/- {2 * init_pred_target_std_khz:.3f} kHz")
print(f"  True physical frequency: {init_true_freq_khz:.3f} kHz (Target: {target_khz:.3f} kHz)")
print(f"  True physical error:     {init_true_rel_error:.3f}%")
```

Plot how the initial surrogate model looks:

```python
param_labels = ['a / mm', 'h / um', 'p / um', 'd / um']
param_names = ['Plate Half-side a', 'Thickness h', 'Pitch p', 'Diameter d']
scale_factors = [1e3, 1e6, 1e6, 1e6]

plt.figure(figsize=(10, 5))

for i in range(4):
    plt.subplot(2, 2, i + 1)
    
    # 1D slice through the initial best surrogate design
    sweep_vals = np.linspace(BOUNDS[i, 0], BOUNDS[i, 1], 250)
    X_slice_phys = np.tile(init_best_params, (250, 1))
    X_slice_phys[:, i] = sweep_vals
    
    X_slice_u = to_unit(X_slice_phys)
    mu_khz, std_khz = gp.predict(X_slice_u, return_std=True)
    
    true_slice_khz = np.array([
        get_freq(row[0], row[1], row[2], row[3], m=1, n=1) / 1e3
        for row in X_slice_phys
    ])
    
    x_plot = sweep_vals * scale_factors[i]
    x_opt = init_best_params[i] * scale_factors[i]
    
    # Model predictions and true 1D physics slice
    plt.plot(x_plot, true_slice_khz, 'k--', linewidth=1.6, label='get_freq()')
    plt.plot(x_plot, mu_khz,  color='darkorange', linewidth=2.0, label='GP Posterior Mean')
    plt.fill_between(
        x_plot,
        mu_khz - 2 * std_khz,
        mu_khz + 2 * std_khz,
        color='orange',
        alpha=0.25,
        label=r"$\pm 2\sigma$"
    )
    
    # Initial 10 sampled points (plotted at their respective parameter coordinate)
    plt.scatter(
        X_train_phys[:, i] * scale_factors[i],
        y_train_khz,
        color='red',
        edgecolors='k',
        s=50,
        zorder=5,
        label=f'Initial Samples ({N_INIT})'
    )
    
    plt.axhline(target_khz, color='crimson', linestyle=':', linewidth=1.5, label=f'Target ({target_khz:.1f} kHz)')
    plt.axvline(x_opt, color='darkgreen', linestyle='-', linewidth=2.0, label='Initial Best Guess')
    
    plt.title(f'Posterior $f \\mid \\mathcal{{D}}$ vs. {param_names[i]}')
    plt.xlabel(param_labels[i])
    plt.ylabel(r'$f_0$ / kHz')
    plt.grid(True, linestyle=':', alpha=0.6)
    
    if i == 1:
        plt.legend(loc='upper left', fontsize=8.5, framealpha=0.85)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# Start Active Learning Loop
# ==============================================================================

for iteration in range(1, MAX_ITER + 1):
    X_train_u = to_unit(X_train_phys)
    gp.fit(X_train_u, y_train_khz)
    
    pred_f_khz, pred_sigma_khz = gp.predict(candidates_u, return_std=True)
    
    acquisition = np.abs(pred_f_khz - target_khz) - 0.3 * pred_sigma_khz
    cand_idx = np.argmin(acquisition)
    query_design = candidates_phys[cand_idx]
    
    query_freq_khz = get_freq(*query_design, m=1, n=1) / 1e3
    curr_rel_error = abs(query_freq_khz - target_khz) / target_khz * 100.0
    
    X_train_phys = np.vstack([X_train_phys, query_design])
    y_train_khz = np.append(y_train_khz, query_freq_khz)
    
    best_params = query_design
    pred_target_khz = pred_f_khz[cand_idx]
    pred_target_std = pred_sigma_khz[cand_idx]
    actual_best_freq = query_freq_khz * 1e3
    rel_error = curr_rel_error
    
    total_samples = len(y_train_khz)
    # print(f"Iteration {iteration:02d} (Total Calls: {total_samples:02d}) -> "
    #       f"Tested: {query_freq_khz:.2f} kHz | "
    #       f"Surrogate Pred: {pred_target_khz:.2f} kHz (std: {pred_target_std:.3f} kHz) | "
    #       f"Rel Error: {curr_rel_error:.3f}%")

    print(f"Iteration {iteration:02d} (Total Calls: {total_samples:02d}) -> "
          f"[a: {query_design[0]*1e3:.3f} mm, h: {query_design[1]*1e6:.2f} um, "
          f"p: {query_design[2]*1e6:.2f} um, d: {query_design[3]*1e6:.2f} um] | "
          f"True: {query_freq_khz:.2f} kHz | "
          f"Surrogate: {pred_target_khz:.2f} +/- {1.96*pred_target_std:.2f} kHz (std: {pred_target_std:.3f} kHz) | "
          f"Rel Error: {curr_rel_error:.3f}%")
    
    if curr_rel_error <= STOPPING_TOL_PCT:
        print(f"\n[Convergence Reached] Relative error below threshold ({STOPPING_TOL_PCT}%).")
        break

print("\n==========================================")
print("Optimal Design Found:")
print(f"  Total Calls Used:     {len(y_train_khz)}")
print(f"  Plate Half-side a:    {best_params[0]*1e6:.3f} um")
print(f"  Thickness h:          {best_params[1]*1e6:.2f} um")
print(f"  Pitch p:              {best_params[2]*1e6:.2f} um")
print(f"  Diameter d:           {best_params[3]*1e6:.2f} um")
print(f"  GP Predicted Freq:    {pred_target_khz:.2f} +/- {1.96*pred_target_std:.2f} kHz")
print(f"  Actual Physics Freq:  {actual_best_freq/1e3:.2f} kHz (Target: {target_khz:.2f} kHz)")
print(f"  Absolute Error:       {rel_error:.3f}%")
print("==========================================")
```

Plot how the final surrogate model looks:

```python
param_labels = ['a / mm', 'h / um', 'p / um', 'd / um']
param_names = ['Plate Half-side a', 'Thickness h', 'Pitch p', 'Diameter d']
scale_factors = [1e3, 1e6, 1e6, 1e6]
plt.figure(figsize=(10,5))

for i in range(4):
    plt.subplot(2, 2, i + 1)
    
    sweep_vals = np.linspace(BOUNDS[i, 0], BOUNDS[i, 1], 250)
    X_slice_phys = np.tile(best_params, (250, 1))
    X_slice_phys[:, i] = sweep_vals
    
    X_slice_u = to_unit(X_slice_phys)
    mu_khz, std_khz = gp.predict(X_slice_u, return_std=True)
    
    true_slice_khz = np.array([
        get_freq(row[0], row[1], row[2], row[3], m=1, n=1) / 1e3
        for row in X_slice_phys
    ])
    
    x_plot = sweep_vals * scale_factors[i]
    x_opt = best_params[i] * scale_factors[i]
    
    plt.plot(x_plot, true_slice_khz, 'k--', linewidth=1.6, label='get_freq()')
    plt.plot(x_plot, mu_khz,  color='darkorange', linewidth=2.0, label='GP Posterior Mean')
    plt.fill_between(
        x_plot,
        mu_khz - 2 * std_khz,
        mu_khz + 2 * std_khz,
        color='orange',
        alpha=0.25,
        label=r"$\pm 2\sigma$"
    )
    plt.axhline(target_khz, color='crimson', linestyle=':', linewidth=1.5, label=f'Target ({target_khz:.1f} kHz)')
    plt.axvline(x_opt, color='darkgreen', linestyle='-', linewidth=2.0, label='Chosen Design')
    
    plt.title(f'Posterior $f \\mid \\mathcal{{D}}$ vs. {param_names[i]}')
    plt.xlabel(param_labels[i])
    plt.ylabel(r'$f_0$ / kHz')
    plt.grid(True, linestyle=':', alpha=0.6)
    
    if i == 0:
        plt.legend(loc='upper right', framealpha=0.85)

plt.tight_layout()
plt.show()

```

Note on the posterior plots:

Each subplot displays a 1D conditional slice of the 4D surrogate by varying a single parameter while freezing the other three at the reference coordinates $x^∗=[a^∗,h^∗,p^∗,d^∗]$ (the green vertical line):  

* Plate Half-side $a$: evaluates $f(a,h^∗,p^∗,d^∗)$
* Thickness $h$: evaluates $f(a^∗,h,p^∗,d^∗)$
* Pitch $p$: evaluates $f(a^∗,h^∗,p,d^∗)$
* Diameter $d$: evaluates $f(a^∗,h^∗,p^∗,d)$

Because the curve and $\pm 2\sigma$ band follow this exact 1D trajectory through the 4D space, the observed training points (dots) will only land directly on the curve if their remaining three parameters happen to match $x^∗$. Otherwise, they appear vertically offset because their full vector $[a,h,p,d]$ sits elsewhere in the parameter space.
