# 1. Introduction

How to work with Jupyter notebook? Most important keycuts:

**Execution**
* `Shift` + `Enter`: Run current cell and move to the next cell (creates a new one if at the end).
* `Ctrl` + `Enter`: Run current cell in-place.
* `Alt` + `Enter`: Run current cell and insert a new blank cell below.

**Command Mode** *(press `Esc` to enter)*
* `A`: Insert cell **A**bove.
* `B`: Insert cell **B**elow.
* `D`: **D**elete current cell.
* `Z`: Undo cell deletion.
* `M`: Convert cell to **M**arkdown.
* `Y`: Convert cell to **C**ode.


**Edit Mode** *(press `Enter` to type inside a cell)*
* `Tab`: Code autocompletion.
* `Shift` + `Tab`: Show docstring / signature tooltip for a function.
